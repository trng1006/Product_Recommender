```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("✅ Đã import thành công các thư viện phân tích dữ liệu.")
```

## 1. Thống kê quy mô tập dữ liệu tương tác khách hàng

```python
# Tổng hợp từ số liệu phân tích phân tán của Spark ETL
total_interactions = 2756101
total_users = 1407580
total_items = 235061

event_distribution = pd.Series({
    'view': 2664312,
    'addtocart': 69332,
    'transaction': 22457
})

print(f"Tổng số lượt tương tác : {total_interactions:,}")
print(f"Tổng số khách hàng (Users): {total_users:,}")
print(f"Tổng số sản phẩm (Items)  : {total_items:,}")
print("\nTỷ lệ các loại hành vi:")
for event, count in event_distribution.items():
    print(f" - {event.ljust(12)}: {count:>10,} ({count / total_interactions * 100:.2f}%)")
```

## 2. Tính toán độ thưa ma trận tương tác (Matrix Sparsity)

```python
total_possible_cells = total_users * total_items
sparsity = (1.0 - (total_interactions / total_possible_cells)) * 100

print(f"Kích thước ma trận tiềm năng: {total_possible_cells:,} ô")
print(f"Độ thưa của ma trận (Sparsity): {sparsity:.5f}%")
print("=> Nhận xét: Ma trận cực kỳ thưa (> 99.99%), lý giải bắt buộc phải dùng thuật toán ALS phân rã ma trận thay cho mạng nơ-ron dày đặc thông thường.")
```

## 3. Trực quan hóa tỷ lệ chuyển đổi hành vi

```python
plt.figure(figsize=(8, 5))
colors = ['#1565C0', '#FFA000', '#2E7D32']
bars = plt.bar(event_distribution.index, event_distribution.values, color=colors)
plt.yscale('log')
plt.title('Phân bổ hành vi tương tác RetailRocket (Thang đo Logarithmic)', fontsize=14, fontweight='bold')
plt.xlabel('Loại hành vi', fontsize=12)
plt.ylabel('Số lượng tương tác (Log scale)', fontsize=12)
for bar in bars:
    yval = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, yval * 1.15, f'{int(yval):,}', ha='center', va='bottom', fontsize=10, fontweight='bold')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
